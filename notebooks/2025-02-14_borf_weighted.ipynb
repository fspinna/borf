```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from fast_borf.weighted.borf_weighted import transform_sax_patterns as transform_sax_patterns_weighted
from fast_borf.bag_of_patterns.borf_new_new_sax import transform_sax_patterns
```

```python
panel = np.random.randn(1_000, 1, 1_000)
timestamps = np.repeat(np.arange(1_000)[None, :], 1_000, axis=0)[:, None, :] + np.random.rand(1_000, 1, 1_000) * 0.1
window_size = 8
word_length = 4
alphabet_size = 3
bins = np.array([-0.67, 0, 0.67])
stride=1
dilation=1
min_window_to_signal_std_ratio=0.0
np.any(timestamps < 0)
```

```python
panel.shape, timestamps.shape
```

```python
%%time
out = transform_sax_patterns_weighted(
    panel=panel,
    panel_timestamps=timestamps,
    window_size=window_size,
    word_length=word_length,
    alphabet_size=alphabet_size,
    dilation=dilation,
    stride=stride,
    min_window_to_signal_std_ratio=min_window_to_signal_std_ratio,
)
```

```python
%%time
out2 = transform_sax_patterns(
    panel=panel,
    window_size=window_size,
    word_length=word_length,
    alphabet_size=alphabet_size,
    dilation=dilation,
    stride=stride,
    min_window_to_signal_std_ratio=min_window_to_signal_std_ratio,
)
```
