```python
SIZE = 100000
arr = np.random.rand(SIZE)
timestamps = np.arange(SIZE)
window_size = 80
word_length = 8
bins = np.array([-0.67, 0, 0.67])
stride=2
dilation=3
min_window_to_signal_std_ratio=0.01
```

```python
%%timeit
sax_weighted(
    arr,
    timestamps,
    window_size,
    word_length,
    bins,
    stride,
    dilation,
    min_window_to_signal_std_ratio,
)
```

```python
%%timeit
sax(
    arr,
    window_size,
    word_length,
    bins,
    stride,
    dilation,
    min_window_to_signal_std_ratio,
)
```

```python
np.allclose(
    sax_weighted(
        arr,
        timestamps,
        window_size,
        word_length,
        bins,
        stride,
        dilation,
        min_window_to_signal_std_ratio,
    ),
    sax(
        arr,
        window_size,
        word_length,
        bins,
        stride,
        dilation,
        min_window_to_signal_std_ratio,
    ),
)
```
